## Wizualizacja separacji klas

Działanie modelu można zobrazować za pomocą położenia wektorów wspierających oraz granic funkcji decyzyjnej.

**Proszę:**
- wypisać wektory wspierające za pomocą funkcji `SVC.support_vectors_`. Oczekiwany wynik to:
```Python
Support vectors:
 [[1.9 0.4]
 [3.  1.1]]
```
* korzystając z funkcji `sns.jointplot`, `plot_support_vectors` i `plot_decision_function` na jednym rysunku narysować:
    - dane
    - wektory wspierające
    - granice funkcji decyzyjnej.

wskazówka: zajrzyj do dokumentacji: https://seaborn.pydata.org/generated/seaborn.JointGrid.html#seaborn.JointGrid

````python
def plot_support_vectors(support_vectors, axis):
  """
  Rysuje wektory wspierające (support vectors) na podanym wykresie.
  Wektory są zaznaczane jako okręgi wokół punktów danych.
  """
  axis.scatter(support_vectors[:,0], support_vectors[:,1],s=100,
               linewidth=1, facecolors='none', edgecolors='k')
  plt.gca().set_aspect('equal', adjustable='box')

def plot_decision_function(model, columns, axis):
  """
  Rysuje granicę decyzyjną oraz marginesy dla dwuwymiarowego liniowego modelu SVM.
  Dodatkowo rysuje wektor wag beta prostopadły do linii podziału.
  """
  xlim = axis.get_xlim()
  ylim = axis.get_ylim()
  # Tworzenie siatki punktów do ewaluacji funkcji decyzyjnej
  xx = np.linspace(xlim[0], xlim[1], 30)
  yy = np.linspace(ylim[0], ylim[1], 30)
  YY, XX = np.meshgrid(yy, xx)
  xy = np.vstack([XX.ravel(), YY.ravel()]).T
  xy = pd.DataFrame(data=xy, columns = columns);
  # Obliczenie wartości funkcji decyzyjnej dla każdego punktu siatki
  Z = model.decision_function(xy).reshape(XX.shape)

  # Obliczenie wektora wag beta na podstawie wektorów wspierających i współczynników dualnych
  beta = np.sum(model.support_vectors_*model.dual_coef_.T,axis=0)
  betaMag = np.sqrt(np.sum(beta*beta))
  intercept = model.intercept_
  yVtx =  (ylim[0] + ylim[1])/2
  xVtx =  (-intercept - beta[1]*yVtx)/beta[0]
  if xVtx<xlim[0] or xVtx>xlim[1]:
    xVtx =  (xlim[0] + xlim[1])/2
    yVtx =  (-intercept - beta[0]*yVtx)/beta[1]
  if xVtx>xlim[0] and xVtx<xlim[1] and yVtx>ylim[0] and yVtx<ylim[1]:
      # Rysowanie wektora beta (kierunku maksymalnego marginesu)
      axis.quiver(xVtx, yVtx, beta[0], beta[1], angles='xy', scale_units="xy", label=r'$\vec{\beta}$')

  # Rysowanie linii marginesu (Z = -1, 1) oraz linii decyzyjnej (Z = 0)
  axis.contour(
      XX, YY, Z, colors="k", levels=[-1, 0, 1], alpha=0.5, linestyles=["--", "-", "--"]
  )
  axis.grid(True)
  plt.gca().set_aspect('equal', adjustable='box')
  axis.legend()
````

````python
sv=model.support_vectors_
print(sv)
ax=sns.jointplot(x=iris_abr['petal length (cm)'],y=iris_abr['petal width (cm)'], hue=iris_abr['label'])
plot_support_vectors(sv, ax.ax_joint)
plot_decision_function(model,["petal length (cm)","petal width (cm)"],ax.ax_joint)
````
